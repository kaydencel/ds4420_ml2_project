# NYC Airbnb — Data Preprocessing

```python
import pandas as pd
from datetime import datetime
import matplotlib.pyplot as plt
import numpy as np
```

## Prepare, explore, and clean data

```python
# Load data
airbnb = pd.read_csv('listings.csv')
airbnb.head(10)
```

```python
# Get info on dataset
airbnb.info()
```

```python
# Check for missing values
missing = airbnb.isnull().sum()
print(missing[missing > 0])
```

```python
# Drop columns with more than 15,000 missing values
airbnb = airbnb.dropna(axis=1, thresh=len(airbnb) - 15000)
```

```python
# Clean Columns

# Change last_review column to datetime 
today = pd.Timestamp.today()
#today = pd.Timestamp(datetime.today().date())
airbnb["last_review"] = pd.to_datetime(airbnb["last_review"], errors="coerce")
 
# Calculate number of days since the listing was last reviewed, if listing has no review, fill with 9999
airbnb["days_since_last_review"] = (today - airbnb["last_review"]).dt.days.fillna(9999)
 
# Clean price column and turn into float
airbnb["price"] = (
    airbnb["price"]
    .astype(str)
    .str.replace(r"[\$,]", "", regex=True)
    .pipe(pd.to_numeric, errors="coerce")
)
 
# Fill NaN values with 0
airbnb["reviews_per_month"] = airbnb["reviews_per_month"].fillna(0)

# Map binary columns, 1 = t, 0 = f or NaN
airbnb['host_identity_verified'] = (airbnb['host_identity_verified'] == "t").astype(int)
airbnb['host_has_profile_pic'] = (airbnb['host_has_profile_pic'] == "t").astype(int)
 
```

## Determine which listings are ghost listings

```python
# Create conditions to classify ghost listing
high_avail = airbnb["availability_365"] >= 360
high_min_nights = airbnb["minimum_nights"] >= 360
num_since_lr = airbnb["days_since_last_review"] >= 360
airbnb["price_percentile"] = airbnb.groupby(["neighbourhood_cleansed", "room_type"])["price"].rank(pct=True)
low_price = airbnb['price_percentile'] < 0.05
sus_host = (airbnb['host_identity_verified'] == 0) & (airbnb['host_has_profile_pic'] == 0)
 
# Assign each listing a score and label to determine if it is a ghost listing
airbnb["ghost_score"] = ( 
                         high_avail.astype(int) + 
                         high_min_nights.astype(int) + 
                         num_since_lr.astype(int) + 
                         low_price.astype(int) +
                         sus_host.astype(int))
 
def assign_label(score):
    if score > 2:
        return 1  # fake listing if ghost score is above 2
    elif score <= 2:
        return 0  # real listing if ghost score is equal or less than 2
 
airbnb["ghost_label"] = airbnb["ghost_score"].apply(assign_label)

counts = airbnb["ghost_label"].value_counts().rename({1: "ghost", 0: "real",})
total  = len(airbnb)

print(counts)
print(f'Total: {total}')
 
```

## MAYBE DELETE Explore which data to log transform

```python
# Price Histograms
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].hist(airbnb["price"], bins=50)
axes[0].set_title("Price Distribution")

price_filtered = airbnb[airbnb["price"] < airbnb["price"].quantile(0.99)]["price"]
axes[1].hist(price_filtered, bins=50)
axes[1].set_title("Price Distribution (Top 99%)")

# Price is skewed right, log transform is a good transformation
airbnb["log_price"] = np.log(airbnb["price"])
axes[2].hist(airbnb["log_price"], bins=50)
axes[2].set_title("Log Price Distribution")

plt.show()
```

```python
# Minimum Nights Histograms - Use original, not good for log transform
fig, axes = plt.subplots(1, 2, figsize=(15, 4))

axes[0].hist(airbnb["minimum_nights"], bins=50)
axes[0].set_title("Minimum Nights Distribution ")

# Minimum nights is skewed right, log transform is not a good transformation
airbnb["log_min_nights"] = np.log(airbnb["minimum_nights"])
axes[1].hist(airbnb["log_min_nights"], bins=50)
axes[1].set_title("Log Minimum Nights Distribution")

plt.show()
```

```python
# Number of Listings a Host has Histograms 
fig, axes = plt.subplots(1, 2, figsize=(15, 4))

axes[0].hist(airbnb["host_listings_count"], bins=50)
axes[0].set_title("Number of Listings a Host has Distribution ")

# Number of Listings a Host is skewed right, log transform is a good transformation
airbnb["log_host_list_count"] = np.log(airbnb["host_listings_count"])
axes[1].hist(airbnb["log_host_list_count"], bins=50)
axes[1].set_title("Log Number of Listings a Host has Distribution")

plt.show()
```

```python
# Reviews per Month Histograms 
fig, axes = plt.subplots(1, 2, figsize=(15, 4))

axes[0].hist(airbnb["reviews_per_month"], bins=50)
axes[0].set_title("Reviews per Month Distribution ")

# Reviews per Month  is skewed right, log transform is a good transformation
airbnb["log_rpm"] = np.log1p(airbnb["reviews_per_month"])
axes[1].hist(airbnb["log_rpm"], bins=50)
axes[1].set_title("Log Reviews per Month  Distribution")

plt.show()
```
