```python
import pandas as pd
import numpy as np
```

```python
airbnb = pd.read_csv('Airbnb_NYC.csv')
```

```python
missing_val_list = []

for values in airbnb:
    print(values, airbnb[values].isnull().sum())
    missing_val_list.append(airbnb[values].isnull().sum())
```

```python
# Drops columns where 1500+ values were missing in them

airbnb = airbnb.dropna(axis=1, thresh=len(airbnb) - 1500)
```
